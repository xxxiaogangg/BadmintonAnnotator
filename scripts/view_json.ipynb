```python
import os
import json
from collections import Counter

files_dir = "/home/gang/gang_study/Data/Badminton_data/InHome/BWF/女单"
```

```python
file_name = "2025年中国羽毛球大师赛-安洗莹vs韩悦.mp4.json"

with open(os.path.join(files_dir, file_name), "r") as f:
    json_data = json.load(f)

events = json_data.get("events", [])

# 1. 统计所有 SHOT 类型事件的 major 和 minor 动作
major_counts = Counter()
minor_counts = Counter()
hands_counts = Counter()

# print(events[0])

for event in events:
    if event.get("type") == "SHOT" or event.get("type") == "RALLY_START":
        details = event.get("details", {})
        hand = details.get("hand")
        major = details.get("major")
        minor = details.get("minor")

        hands_counts[hand] += 1
        major_counts[major] += 1
        minor_counts[minor] += 1

print(hands_counts)
print(major_counts)
print(minor_counts)
```
